```sql
CREATE TABLE IF NOT EXISTS employee (
    id INT,
    name STRING,
    department STRING,
    salary INT
)
USING DELTA;

INSERT INTO employee (id, name, department, salary) VALUES
(1, 'Alice', 'HR', 70000),
(2, 'Bob', 'Engineering', 90000),
(3, 'Charlie', 'Sales', 65000),
(4, 'Diana', 'Marketing', 72000),
(5, 'Eve', 'Engineering', 85000),
(6, 'Frank', 'Engineering', 88000),
(7, 'Grace', 'Sales', 68000),
(8, 'Hannah', 'HR', 75000),
(9, 'Ivan', 'Marketing', 78000),
(10,'Jack', 'Engineering', 92000);
```

```sql
-- Aggregation & Grouping (31-40)
--Summarize and analyze data with aggregate functions.
--1.Count rows for each group.
select department, count(*) from employee group by department
```

```sql
--2. calculate average salary per department 
select department, avg(salary) from employee group by department
```

```sql
--3.Find the total salary expenditure for each department 
select department, sum(salary) from employee group by department
```

```sql
--4.find the maximum salary in each department 
select department, max(salary) from employee group by department
```

```sql
--5. find min salary from each department 
select department, min(salary) from employee group by department
```

```sql
--6. Filter groups using having 
select department, avg(salary) from employee group by department having avg(salary) > 80000
```

```sql
--7.count employees in departments with more than 5 employees-- 
select department, count(*) from employee group by department having count(*) >3
```

```sql
--8.filter and then group
select department, avg(salary) from employee where salary >80000 group by department
```

```sql

--9.Group by multiple columns.
SELECT department, AVG(salary) FROM employee GROUP BY department
```

```sql
--10. Find the first and last name of the highest earner in each department.
SELECT department, MAX(salary) FROM employee GROUP BY department;
```

```sql
--10.filter by salary range
select * from employee where salary between 70000 and 80000
```

```sql
--get the employee with the highest salary select * from employee where salary = (select max(salary) from employee)
```
