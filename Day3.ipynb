```python
#5.Comparing two strings for anagrams
str1 = "Listen"
str2 = "Silent"
str1 = str1.replace("","").upper()
str2 = str2.replace("","").upper()
if sorted(str1) == sorted(str2):
    print("The strings are anagrams")
else:
  print("False")
```

```python
#6.checking for palindrome using extended slicing technique
str1  = "Kayak".lower()
if str1 == str1[::-1]:
    print("The string is a palindrome")
else:
    print("The string is not a palindrome")
```

```python
#8.Counting Digits, letters and spaces in string
new_string = "ajay krishna 144"

digits_count = 0
letters_count = 0
spaces_count = 0

for char in new_string:
    if char.isdigit():
        digits_count += 1
    elif char.isalpha():
        letters_count += 1
    elif char.isspace():
        spaces_count += 1

print(f"Digits: {digits_count}")
print(f"Letters: {letters_count}")
print(f"Spaces: {spaces_count}")
```

```python
def count_special_characters(string):
  sp_char = '@22@#$%(&&%%)'
  count = 0
  for i in string:
    if i in sp_char:
      count += 1
  return count
print(count)

```

```python
#10.removing all white space in a string
import re
string = "c o d e"
spaces = re.compile(r'\s+')
result = re.sub(spaces, "", string)
print(result)
```
