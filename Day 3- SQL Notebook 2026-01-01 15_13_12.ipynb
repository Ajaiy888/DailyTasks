```sql
CREATE TABLE IF NOT EXISTS employee (
    id INT,
    name STRING,
    department STRING,
    salary INT
)
USING DELTA;

INSERT INTO employee (id, name, department, salary) VALUES
(1, 'Alice', 'HR', 70000),
(2, 'Bob', 'Engineering', 90000),
(3, 'Charlie', 'Sales', 65000),
(4, 'Diana', 'Marketing', 72000),
(5, 'Eve', 'Engineering', 85000),
(6, 'Frank', 'Engineering', 88000),
(7, 'Grace', 'Sales', 68000),
(8, 'Hannah', 'HR', 75000),
(9, 'Ivan', 'Marketing', 78000),
(10,'Jack', 'Engineering', 92000);
```

```sql
--1.sorting and limiting
--control the order of your results and limit the number of rows
--1.sort results in ascending order
select * from employee order by name asc;
```

```sql
--2.sort results in descending order
select * from employee order by salary desc
```

```sql
--3.sort by multiple columns
select * from employee order by department asc,salary asc
```

```sql
--4.get the top 10 highest paid employees 
select * from employee order by salary desc limit 3
```

```sql
--5.get the latest 5 hired employees 
select * from employee order by order_date desc limit 5
```

```sql
--6. combine distinct and order by 
select distinct department from employee order by department desc
```

```sql
--7.fetch a specific range of rows(ex 5-10)
select * from employee order by id limit 5 offset 5
```

```sql
--8.get the employee with the highest salary
select * from employee order by salary desc limit 1
```

```sql
--9.get the second-highest salary
select * from employee where salary = (select max(salary) from employee where salary < (select max(salary) from employee))
```

```sql
--10. Order by a caluculated field
select name, department , salary, salary * 1.1 as new_salary from employee order by new_salary desc
```

```sql
--10.filter by salary range
select * from employee where salary between 70000 and 80000
```

```sql
--get the employee with the highest salary select * from employee where salary = (select max(salary) from employee)
```
