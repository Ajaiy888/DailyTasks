```python
#create pyspark session,
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Spark DataFrames").getOrCreate()
```

```python
from pyspark.sql import SparkSession
from pyspark.sql import Row
from pyspark.sql.types import StructType, StructField, IntegerType, StringType, DoubleType
from pyspark.sql.functions import *
```

```python
data = [Row(1, "John", 30, "Sales", 50000.0),
 Row(2, "Alice", 28, "Marketing", 60000.0),
 Row(3, "Bob", 32, "Finance", 55000.0),
 Row(4, "Sam", 35, "Sales", 52000.0),
 Row(5, "Emily", 29, "Marketing", 62000.0),
 Row(6, "Mike", 31, "Finance", 58000.0),
 Row(7, "Lucy", 33, "Sales", 54000.0),
 Row(8, "Tom", 37, "Marketing", 65000.0),
 Row(9, "Sara", 30, "Finance", 56000.0),
 Row(10, "Jack", 34, "Sales", 59000.0)]
schema = StructType([
 StructField("id", IntegerType(), True),
 StructField("name", StringType(), True),
 StructField("age", IntegerType(), True),
 StructField("department", StringType(), True),
 StructField("salary", DoubleType(), True)
])
employeeDF = spark.createDataFrame(data, schema)
employeeDF.createOrReplaceTempView("employee")
employeeDF.show()
employeeDF.printSchema()    
```

```python
# show the dataframe
employeeDF.show()
```

```python
#display the dataframe
display(employeeDF)
```

```python
#calculate the average salary for each department
avgsalbydept = employeeDF.groupBy("department").agg(avg("salary").alias("avg_salary"))
avgsalbydept.show()
```

```python
#Add a new column named "bonus" that is 10% of the salary for all employees
employeeDF = employeeDF.withColumn("bonus", col("salary") * 0.1)
employeeDF.show()
```

```python
#Group the data by department and find the employee with the highest salary in each department
from pyspark.sql.functions import max
highestSalByDept = employeeDF.groupBy("department").agg(max("salary").alias("salary"))
highestSalByDept.show() 
```
